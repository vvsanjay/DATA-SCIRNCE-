```python

import pandas as pd
import numpy as np

data = {
    "emp_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": [
        "Alice Johnson", "Bob Smith",   "Carol White", "David Brown",
        "Eva Martinez", "Frank Lee",   "Grace Kim",   "Henry Wilson"
    ],
    "designation": [
        "Manager",  "Developer", np.nan,     "Analyst",
        "Designer", np.nan,      "Tester",   "DevOps"
    ],
    "years_of_experience": [8,  3,  5,  2,  6,  1,  4,  7],
    "age":                 [35, 27, 30, 25, 32, 23, 28, 34],
    "salary":              [12000, 8500, 9800, 7500, 11500, 6000, 9000, 15000],
}

df = pd.DataFrame(data)

print("=" * 60)
print("           ORIGINAL EMPLOYEE RECORDS")
print("=" * 60)
print(df.to_string(index=False))

df_filtered = df[df["salary"] <= 10000].copy()

print("\n" + "=" * 60)
print("  AFTER dropping rows with Salary > 10,000")
print("=" * 60)
print(df_filtered.to_string(index=False))
print(f"\n  Rows dropped : {len(df) - len(df_filtered)}  "
      f"(emp_ids: {df[df['salary'] > 10000]['emp_id'].tolist()})")

df_filtered["designation"] = df_filtered["designation"].fillna("Not Assigned")

print("\n" + "=" * 60)
print("  AFTER handling NA in Designation")
print("=" * 60)
print(df_filtered.to_string(index=False))

df_filtered.rename(columns={"years_of_experience": "exp"}, inplace=True)

print("\n" + "=" * 60)
print("  AFTER renaming 'years_of_experience' → 'exp'")
print("=" * 60)
print(df_filtered.to_string(index=False))

df_filtered["age"] = df_filtered["age"] + 3

print("\n" + "=" * 60)
print("  AFTER incrementing Age by 3")
print("=" * 60)
print(df_filtered.to_string(index=False))

print("\n" + "=" * 60)
print("              FINAL PROCESSED DATAFRAME")
print("=" * 60)
print(df_filtered.to_string(index=False))

print("\n── DataFrame Info ──────────────────────────────────────")
print(f"  Total records  : {len(df_filtered)}")
print(f"  Columns        : {list(df_filtered.columns)}")
print(f"  Null values    : {df_filtered.isnull().sum().to_dict()}")
print("=" * 60)
```
